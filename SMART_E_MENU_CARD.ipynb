```python
# E-Restaurant Ordering System (Colab simulation of Arduino code)

itemNames = ["IDLY", "DOSA", "PULAO", "FRIED RICE", "VEG MEALS", "CHICKEN BIRY"]
prices = [100, 150, 200, 250, 300, 350]
itemQty = [0, 0, 0, 0, 0, 0]

totalItems = 0
totalCost = 0

def show_restaurant_menu():
    print("====== RESTAURANT MENU ======")
    for i in range(6):
        print(f"{i+1}. {itemNames[i]} - Rs.{prices[i]}")
    print("A-F REMOVE | 7 ORDER")
    print("=============================")

def add_item(i):
    global totalItems, totalCost
    itemQty[i] += 1
    totalItems += 1
    totalCost += prices[i]
    print(f"{itemNames[i]} ADDED, QTY: {itemQty[i]}")

def remove_item(i):
    global totalItems, totalCost
    if itemQty[i] > 0:
        itemQty[i] -= 1
        totalItems -= 1
        totalCost -= prices[i]
        print(f"{itemNames[i]} REMOVED")
    else:
        print(f"{itemNames[i]} - Nothing to remove")

def show_order_summary():
    print("----- ORDER SUMMARY -----")
    for i in range(6):
        if itemQty[i] > 0:
            print(f"{itemNames[i]} x {itemQty[i]} = Rs.{itemQty[i]*prices[i]}")
    print(f"TOTAL ITEMS: {totalItems}")
    print(f"TOTAL COST : Rs.{totalCost}")

def ask_payment():
    print("SELECT PAYMENT")
    print("1. CASH")
    print("2. CARD / UPI")
    while True:
        choice = input("Enter 1 or 2: ").strip()
        if choice in ['1', '2']:
            return int(choice)
        print("Invalid choice, try again.")

def confirm_payment(mode):
    mode_str = "CASH" if mode == 1 else "CARD/UPI"
    print(f"{mode_str} SELECTED")
    while True:
        c = input("CONFIRM Y/N: ").strip().upper()
        if c == 'Y':
            return True
        elif c == 'N':
            return False
        print("Invalid input, enter Y or N.")

def place_order(mode):
    global totalItems, totalCost
    print("ORDER PLACED SUCCESS")
    print(f"PAYMENT: {'CASH' if mode == 1 else 'CARD/UPI'}")
    print("THANK YOU")

    for i in range(6):
        itemQty[i] = 0
    totalItems = 0
    totalCost = 0


# ---------------- MAIN LOOP ----------------
print("WELCOME TO")
print("E-RESTAURANT")
print()

show_restaurant_menu()

while True:
    cmd = input("Enter choice (1-6 add, A-F remove, 7 order, Q quit): ").strip()

    if not cmd:
        continue

    if cmd.upper() == 'Q':
        print("Exiting...")
        break

    if cmd.isdigit() and cmd in '123456':
        add_item(int(cmd) - 1)

    elif cmd.upper() in 'ABCDEF':
        remove_item(ord(cmd.upper()) - ord('A'))

    elif cmd == '7':
        if totalItems > 0:
            show_order_summary()
            mode = ask_payment()
            if confirm_payment(mode):
                place_order(mode)
                show_restaurant_menu()
            else:
                print("PAYMENT CANCELLED")
                show_restaurant_menu()
        else:
            print("NO ITEMS SELECTED")

    else:
        print("Invalid input")
```
